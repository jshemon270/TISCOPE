```python
from cliffs_delta import cliffs_delta
results = []
for features in central_loci.columns:
    d, res = cliffs_delta(peri_loci[features].to_list(),meta_loci[features].to_list())
    
    results.append({
        'feature': features,
        'delta': d,
        'interpretation': res
    })
results = pd.DataFrame(results)
```

```python
import matplotlib.pyplot as plt

color_map = {
    "negligible": "gray",
    "small": "orange",
    "medium": "red",
    "large": "darkred"
}

df = results.copy()
df = df.sort_values("delta")

colors = df["interpretation"].map(color_map)

plt.figure(figsize=(10, 6))

plt.barh(df["feature"], df["delta"], color=colors)

plt.axvline(0, color="black", linewidth=1)

plt.xlabel("Cliff's Delta")
plt.title("Effect Sizes")

plt.tight_layout()
plt.show()
```
