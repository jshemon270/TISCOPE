```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
# X is your feature matrix, you got it before
X1 = pd.read_csv("data_frame_w_intensities_normalized.csv", delimiter=",", quotechar='|')
df_intensity = X1[[c for c in X1.columns if c.endswith('normalized')]]
feature_names = df_intensity.columns

numeric_df = df_intensity.select_dtypes(include=[np.number])

# 1) standardize
scaler = StandardScaler()
X_std = scaler.fit_transform(numeric_df)

# 2) fit k-means
k = 9 ### However many clusters you have
kmeans = KMeans(n_clusters=k, random_state=42)
kmeans.fit(X_std)

labels = kmeans.labels_
centroids = kmeans.cluster_centers_ # shape (k, n_features)
```

```python
# absolute centroid values = how many std deviations from mean per feature
abs_centroids = np.abs(centroids)  # already in standardized space

# turn into a DataFrame for readability
imp_df = pd.DataFrame(abs_centroids, columns=feature_names)
imp_df.index = [f"cluster_{i}" for i in range(k)]

# show top features per cluster
top_n =5
for c in range(k):
    print(f"\nCluster {c}")
    print(imp_df.loc[f"cluster_{c}"].sort_values(ascending=False).head(top_n))
```

```python
centroid_df = pd.DataFrame(centroids, columns=feature_names, index=[f"cluster_{i}" for i in range(k)])

for c in range(k):
    print(f"\nCluster {c} (with sign)")
    print(centroid_df.loc[f"cluster_{c}"].sort_values(key=lambda s: np.abs(s), ascending=False).head(top_n))
```
